**Import Pandas and the Data**

**(It is loaded into your Google Colab Working Directory as "Ecommerce Purchases")**

```python
import pandas as pd
df = pd.read_csv(
    "Ecommerce Purchases",
    sep=",",
    quotechar='"',
    engine="python"
)
```

**Check the head of the DataFrame.**

```python
df.head()
```

**How many rows and columns are there?**

```python
num_rows = df.shape[0]
print(num_rows)

num_columns = df.shape[1]
print(num_columns)
```

**What is the average Purchase Price?**

```python
avg_price = df['Purchase Price'].mean()
print(avg_price)
```

**What were the highest and lowest purchase prices?**

```python
max_price = df['Purchase Price'].max()
print(max_price)
min_price = df['Purchase Price'].min()
print(min_price)
```

**How many people have English 'en' as their Language of choice on the website?**

```python
count_people_en = df[df['Language'].str.lower() == 'en'].shape[0]
print(count_people_en)
```

**How many people have the job title of "Lawyer"?**


```python
count_job_lawyer = df[df['Job'].str.lower() == 'lawyer'].shape[0]
print(count_job_lawyer)
```

**How many people made the purchase during the AM and how many people made the purchase during PM?**

**(Hint: Check out [value_counts()](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.value_counts.html))**

```python
count_people_purchase_am = df[df['AM or PM'].str.lower() == 'am'].shape[0]
print(count_people_purchase_am)

count_people_purchase_pm = df[df['AM or PM'].str.lower() == 'pm'].shape[0]
print(count_people_purchase_pm)
```

**What are the 5 most common Job Titles?**

```python
most_5_common_jobs = df['Job'].value_counts().head(5)
print(most_5_common_jobs)
```

**Someone made a purchase that came from Lot: "90 WT" , what was the Purchase Price for this transaction?**

```python
purchase_price = df.loc[df['Lot'] == "90 WT", 'Purchase Price']
print(purchase_price)
```

**What is the email of the person with the following Credit Card Number: 4926535242672853**

```python
result = df.loc[df["Credit Card"].astype(str) == "4926535242672853", "Email"].values[0]
print(result)
```

**How many people have American Express as their Credit Card Provider *and* made a purchase above $95?**

```python
result = df.loc[(df['CC Provider'] == 'American Express') & (df['Purchase Price'] > 95)]
print(len(result))
```

**How many people have a credit card that expires in 2025?**

```python
df["CC Exp Year"] = pd.to_datetime(df["CC Exp Date"], format="%m/%y", errors="coerce").dt.year
num_people_card_expires_2025 = (df["CC Exp Year"] == 2025).sum()
print(num_people_card_expires_2025)
```

**What are the top 5 most popular email providers/hosts (e.g. gmail.com, yahoo.com, etc...)**

```python
num_5_top_emails_providers = df["Email"].str.split("@").str[1].value_counts().head(5)
print(num_5_top_emails_providers)
```
