**Import pandas as pd. Import numpy as np.**

```python
import pandas as pd
import numpy as np
```

**Read Salaries.csv as a dataframe called sal.**

```python
sal = pd.read_csv('Salaries.csv')
```

**Check the head of the DataFrame.**

```python
sal.head()
```

**Use the .info() method to find out how many entries there are.**

```python
sal.info()
```

**What is the average BasePay?**

```python
sal['BasePayNumeric'] = pd.to_numeric(sal['BasePay'], errors='coerce')
avg_basepay = sal['BasePayNumeric'].mean()
print(avg_basepay)
```

**What is the highest amount of OvertimePay in the dataset ?**

```python
sal['highest_amount_overtime_payNumeric'] = pd.to_numeric(sal['OvertimePay'], errors='coerce')
highest_amount_overtime_pay = sal['highest_amount_overtime_payNumeric'].mean()
print(highest_amount_overtime_pay)
```

**What is the job title of  JOSEPH DRISCOLL ? Note: Use all caps, otherwise you may get an answer that doesn't match up (there is also a lowercase Joseph Driscoll).**

```python
employeeNameInput = "Joseph Driscoll" #JOSEPH DRISCOLL
sal[sal['EmployeeName'].str.lower() == employeeNameInput.lower()]['JobTitle']
```

**How much does JOSEPH DRISCOLL make (including benefits)?**

```python
employeeNameInput = "JOSEPH DRISCOLL"
sal.loc[sal['EmployeeName'].str.lower() == employeeNameInput.lower(), ['Year', 'TotalPayBenefits']]
```

**What is the name of highest paid person (including benefits)?**

```python
sal.loc[sal['TotalPayBenefits'].idxmax(), ['EmployeeName', 'TotalPayBenefits']]
```

**What is the name of lowest paid person (including benefits)? Do you notice something strange about how much he or she is paid?**

```python
sal.loc[sal['TotalPayBenefits'].idxmin(), ['EmployeeName', 'TotalPayBenefits']]
```

**What was the average (mean) BasePay of all employees per year? (2011-2014)?**

```python
df_rango_anios = sal[(sal['Year'] >= 2011) & (sal['Year'] <= 2014)]
df_rango_anios.groupby('Year')['BasePayNumeric'].mean()
```

**How many unique job titles are there?**

```python
num_unique_jobs = sal['JobTitle'].nunique()
print(num_unique_jobs)
```

**What are the top 5 most common jobs?**

```python
top_5_jobs = sal['JobTitle'].value_counts().head(5)
print(top_5_jobs)
```

**How many Job Titles were represented by only one person in 2013? (e.g. Job Titles with only one occurence in 2013?)**

```python
job_titles_2013 = sal[sal['Year'] == 2013]['JobTitle'].value_counts()
#print(job_titles_2013)
num_job_titles_2013 = (job_titles_2013 == 1).sum()
print(num_job_titles_2013)
```

**How many people have the word Chief in their job title? (This is pretty tricky)**

```python
word_search = 'chief'
num_people_word_jobtitle = sal["JobTitle"].str.contains("chief", case=False, na=False).sum()
print(num_people_word_jobtitle)
```

**Bonus: Is there a correlation between length of the Job Title string and Salary?**

```python
sal['title_len'] = sal['JobTitle'].str.len()
correlation = sal['title_len'].corr(sal['TotalPayBenefits'])
print(correlation)
```
